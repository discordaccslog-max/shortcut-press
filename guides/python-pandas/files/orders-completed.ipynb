## 3. Clean

```python
df = df.drop_duplicates()
df["region"] = df["region"].str.strip().str.title()
df["units"] = df["units"].fillna(0).astype(int)
df["date"] = pd.to_datetime(df["date"])
df["region"].value_counts()
```

## 4. Calculate and group

```python
df["revenue"] = df["units"] * df["unit_price"]
round(df["revenue"].sum(), 2)
```

```python
df.groupby("region")["revenue"].agg(["sum", "mean", "count"]).round(2).sort_values("sum", ascending=False)
```

## 5. Combine

```python
regions = pd.read_csv("regions.csv")
q1 = df[df["date"] < "2026-04-01"].groupby("region", as_index=False)["revenue"].sum()
q1 = q1.merge(regions, on="region", how="left")
q1["vs_target"] = q1["revenue"] - q1["quarter_target"]
q1
```

## 6. Time and reshaping

```python
monthly = df.groupby(df["date"].dt.to_period("M"))["revenue"].sum()
monthly
```

```python
df.pivot_table(index="region", columns=df["date"].dt.month, values="revenue", aggfunc="sum").round(2)
```

## 7. Chart and export

```python
monthly.plot(kind="bar", title="Revenue by month")
```

```python
df.to_csv("orders-clean.csv", index=False)
q1.to_excel("q1-vs-target.xlsx", index=False)
```
